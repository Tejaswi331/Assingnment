```python
import numpy as np
marks=np.array([25,78,90,76,54,70,86,98])
print("Average :",np.mean(marks))
print("Highest :",np.max(marks))
print("Lowest  :",np.min(marks))
top_students=marks[marks>70]
print(top_students)
```

```python
import numpy as np
import matplotlib.pyplot as plt
x=np.array([1,2,3,4,5,6])
y=np.array([10,20,30,40,50,60])
plt.plot(x,y)
plt.title("Line chart")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
x=np.array([1,2,3,4,5,6])
y=np.array([10,20,30,40,50,60])
plt.bar(x,y)
plt.title("Bar chart")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()
```
